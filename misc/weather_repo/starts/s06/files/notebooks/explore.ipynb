# The hourly log of weather.csv

```python
lines = open("../data/raw/weather.csv").read().splitlines()
len(lines)
```

```python
lines[1]
```

```python
lines[-1]
```

```python
for line in lines[1:4]:
    print(line)
```

```python
for line in lines[1:4]:
    temp = float(line.split(",")[1])
    if temp < 25:
        print(temp, "below 25")
    else:
        print(temp, "25 or warmer")
```

```python
total = 0.0
count = 0
```

```python
for line in lines[1:]:
    temp = float(line.split(",")[1])
    if temp == -999:
        continue
    total = total + temp
    count = count + 1
```

```python
count, total / count
```

```python
f"{count} hours, mean {total / count:.1f} °C"
```

```python
first = (lines[1][:16], float(lines[1].split(",")[1]))
first
```

```python
hours = {}
for line in lines[1:]:
    day = line[:10]
    temp = float(line.split(",")[1])
    if temp != -999:
        hours[day] = hours.get(day, 0) + 1
```
